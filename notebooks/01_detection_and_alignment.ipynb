## 1. Synthetic Face & Landmark Test Data Generation
Creating a synthetic image with face structures to verify detection and landmark alignment logic.

```python
# Create synthetic face canvas (400 x 400)
img = np.full((400, 400, 3), 220, dtype=np.uint8)

# Face head shape
cv2.ellipse(img, (200, 200), (90, 120), 0, 0, 360, (180, 190, 210), -1)
# Eyes
cv2.circle(img, (160, 170), 12, (50, 50, 50), -1)
cv2.circle(img, (240, 170), 12, (50, 50, 50), -1)
# Nose
cv2.circle(img, (200, 210), 8, (60, 60, 60), -1)
# Mouth corners
cv2.circle(img, (165, 255), 7, (40, 40, 180), -1)
cv2.circle(img, (235, 255), 7, (40, 40, 180), -1)

landmarks = np.array([
    [160.0, 170.0],  # Left Eye
    [240.0, 170.0],  # Right Eye
    [200.0, 210.0],  # Nose
    [165.0, 255.0],  # Left Mouth Corner
    [235.0, 255.0]   # Right Mouth Corner
], dtype=np.float32)

plt.figure(figsize=(5, 5))
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Synthetic Input Image")
plt.axis("off")
plt.show()
```

## 2. 5-Point Landmark Alignment ($112 \times 112$)
Using `FaceAligner` to warp and crop the face into standard $112 \times 112$ recognition input format.

```python
aligner = FaceAligner(crop_size=112)
aligned_face = aligner.align(img, landmarks)

print(f"Aligned face shape: {aligned_face.shape}")

fig, axes = plt.subplots(1, 2, figsize=(8, 4))
axes[0].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
axes[0].scatter(landmarks[:, 0], landmarks[:, 1], c="red", s=40, label="Landmarks")
axes[0].set_title("Original Image + 5 Landmarks")
axes[0].legend()
axes[0].axis("off")

axes[1].imshow(cv2.cvtColor(aligned_face, cv2.COLOR_BGR2RGB))
axes[1].set_title("Aligned Face Crop (112 x 112)")
axes[1].axis("off")

plt.tight_layout()
plt.show()
```
